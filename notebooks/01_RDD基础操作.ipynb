```python

from pyspark.sql import SparkSession
sparkBuilder = SparkSession.builder\
    .appName("rddDemo")\
    .master("local[1]")\
    .getOrCreate()
sc = sparkBuilder.sparkContext
```

```python
#创建一个1-10数组的RDD，将所有元素*2形成新的Rdd map
#data
rdd = sc.parallelize(range(1,11))
print(rdd.collect())
rdd1 = rdd.map(lambda x: x*2).collect()
print(rdd1)
```

```python
#Todd 创建10-20的RDD，使用mapPartitions将所有元素*2
range(10,21)
rdd = sc.parallelize(range(10,21))
print(rdd.collect())
rdd1 = rdd.mapPartitions(lambda iter:[x*2 for x in iter])
print(rdd1.collect())
#
```

```python
rdd1 = sc.parallelize( [1,2,3,4,5])
rdd2 = sc.parallelize([4,5,6,7,8])
rdd3 = rdd1.union(rdd2)
print(rdd3.collect())
rdd4 = rdd1.intersection(rdd2)
print(rdd4.collect())
```

```python

#1.将数据转换成（姓名，分数）格式
rdd = sc.textFile("../data/score.txt")
# map：每行分割成列表，保留一行对应一个列表；不要flatMap
rdd1 = rdd.map(lambda x:x.split(",")) \
          .map(lambda x : (x[0], float(x[2])))
print(rdd1.collect())
```

```python
#2.计算每个学生总分数
rdd2 = rdd1.reduceByKey(lambda x,y:x+y)
print(rdd2.collect())
rdd3 = rdd2.sortBy(lambda x:x[1], ascending = False)
print(rdd3.collect())
```
